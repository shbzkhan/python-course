### Questions of while loop

1. print all the numbers which are divisible by 3 and 5, from 1 to 100

```python
i=1
while i<=100:
    if i%3 == 0 and i%5 == 0:
        print(i, end=" ")
    i +=1
```

2. Sum of all the numbers from 1 to 100

```python
total_sum = 0
i = 1
while i<=100:
    total_sum += i
    i +=1
print(f"Total sum: {total_sum}")
```

3. Sum of all numbers from 1 to 100 divisible by 2 and 7.

```python
total_sum_of_2_7 = 0
i = 1
while i<=100:
    if i%2 ==0 and i%7 == 0:
        total_sum_of_2_7 +=i
    i+= 1
print(f"Total sum of divisible by 2 and 7: {total_sum_of_2_7}")
```
