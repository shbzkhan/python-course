1. Take a person's age and whether they have a valid ID(ture/false) as input. They can enter a venue only if they are 18 or older AND have a valid ID. Print the appropriate message.

```python
age = int(input("Enter your age: "))
valid_id = input("Do you have a valid ID? (True/False): ").strip().capitalize()
print(valid_id)
if age >= 18 and valid_id == "True":
    print("They can enter in venue")
else:
    print("They can not enter in venue")
```
